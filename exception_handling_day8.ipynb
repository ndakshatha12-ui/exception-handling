## Common Python exceptions

```python
# ValueError - wrong value for the operation
try:
    int("abc")
except ValueError as e:
    print("ValueError:", e)
```

```python
# ZeroDivisionError
try:
    10 / 0
except ZeroDivisionError as e:
    print("ZeroDivisionError:", e)
```

```python
# TypeError - wrong type for the operation
try:
    "5" + 5
except TypeError as e:
    print("TypeError:", e)
```

```python
# IndexError - list index out of range
try:
    lst = [1, 2, 3]
    lst[10]
except IndexError as e:
    print("IndexError:", e)
```

```python
# KeyError - missing dictionary key
try:
    d = {"a": 1}
    d["b"]
except KeyError as e:
    print("KeyError:", e)
```

## A program using try-except-else-finally

```python
def divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero")
        return None
    else:
        print("Division successful")
        return result
    finally:
        print("Operation attempted")

print(divide(10, 2))
print(divide(10, 0))
```

## Catching multiple exception types in one block

```python
def safe_convert(val):
    try:
        return int(val)
    except (ValueError, TypeError) as e:
        print("Conversion failed:", e)
        return None

print(safe_convert("42"))
print(safe_convert("abc"))
print(safe_convert(None))
```

## Data conversion program that safely handles invalid values

```python
raw_data = ["23", "45", "abc", "67", "", "12.5", None]

cleaned = []
for val in raw_data:
    try:
        num = int(val)
    except (ValueError, TypeError):
        num = None
    cleaned.append(num)

print(cleaned)
```

```python
# Only the successfully converted values are used in the calculation
valid = [v for v in cleaned if v is not None]
print("Valid values:", valid)
print("Average:", sum(valid) / len(valid))
```

## Handling multiple fields with different exception types

```python
def process_record(record):
    try:
        age = int(record["age"])
        name = record["name"]
    except KeyError as e:
        print("Missing field:", e)
        return None
    except ValueError:
        print("Invalid age value")
        return None
    return {"name": name, "age": age}

print(process_record({"name": "Asha", "age": "25"}))
print(process_record({"name": "Raj"}))
print(process_record({"name": "Meena", "age": "abc"}))
```

## Interview questions

**Why is exception handling important?**
Real data is never perfectly clean — missing fields, wrong types, unexpected values. Without exception handling, one bad row crashes the whole program. With `try/except`, the program can skip, log, or substitute a default for the bad value and keep processing everything else.

**What is the difference between ValueError and TypeError?**
`ValueError` happens when a value has the right type but an inappropriate value for the operation — like `int("abc")`, where a string was expected but its content isn't a valid number. `TypeError` happens when the type itself is wrong for the operation — like adding a string and an integer directly with `+`, where no amount of a different value would fix it.

**What is the purpose of finally?**
Code inside `finally` always runs, whether the `try` block succeeded, failed, or even returned early. It's used for cleanup that must happen no matter what — closing a file, releasing a resource, or logging that an operation was attempted.
